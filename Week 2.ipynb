```python
from scipy.stats import mannwhitneyu

active = customer_df[
    customer_df["Activity_Status"] == "Active"
]

inactive = customer_df[
    customer_df["Activity_Status"] == "Inactive"
]

metrics = [
    "Frequency",
    "Monetary",
    "Total_Quantity",
    "Total_Profit",
    "Avg_Discount",
    "Average_Order_Value"
]

results = []

for metric in metrics:
    stat, p_value = mannwhitneyu(
        active[metric],
        inactive[metric],
        alternative="two-sided"
    )
    
    results.append({
        "Metric": metric,
        "U_Statistic": stat,
        "P_Value": p_value
    })

test_results = pd.DataFrame(results)

test_results
```

```python
customer_profile = df.groupby("Customer ID").agg(
    Category=("Category", lambda x: x.mode()[0]),
    Segment=("Segment", lambda x: x.mode()[0]),
    Market=("Market", lambda x: x.mode()[0])
).reset_index()

customer_test_df = customer_df.merge(
    customer_profile,
    on="Customer ID",
    how="left"
)
```

```python
from scipy.stats import chi2_contingency

categorical_vars = [
    "Category",
    "Segment",
    "Market"
]

chi_results = []

for var in categorical_vars:

    table = pd.crosstab(
        customer_test_df[var],
        customer_test_df["Activity_Status"]
    )

    chi2, p_value, dof, expected = chi2_contingency(table)

    chi_results.append({
        "Variable": var,
        "Chi2_Statistic": chi2,
        "P_Value": p_value,
        "Degrees_of_Freedom": dof
    })

chi_square_results = pd.DataFrame(chi_results)

chi_square_results
```
